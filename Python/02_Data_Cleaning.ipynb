```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    r"C:\Users\asus\OneDrive\Documents\Ecommerce-Sales-Analytics\Data\realistic_e_commerce_sales_data.csv"
)
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()
```

```python
text_columns = [
    "Customer ID",
    "Gender",
    "Region",
    "Product Name",
    "Category",
    "Shipping Status"
]

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)
```

```python
df["Order Date"].isnull().sum()
```

```python
df["Region"] = df["Region"].fillna("Unknown")
```

```python
median_age = df["Age"].median()

median_age
```

```python
df["Age"] = df["Age"].fillna(median_age)
```

```python
df["Age"].isnull().sum()
```

```python
df["Shipping Status"] = df["Shipping Status"].fillna("Unknown")
```

```python
df["Shipping Status"].isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df["Calculated Total"] = (
    df["Unit Price"] * df["Quantity"]
)
```

```python
df["Difference"] = (
    df["Total Price"] - df["Calculated Total"]
)
```

```python
df["Difference"].value_counts()
```

```python
df["Price Check"] = np.where(
    np.isclose(
        df["Total Price"],
        df["Calculated Total"]
    ),
    "Valid",
    "Review"
)
```

```python
df["Price Check"].value_counts()
```

```python
df[df["Price Check"] == "Review"]
```

```python
df.loc[
    df["Price Check"] == "Review",
    [
        "Customer ID",
        "Product Name",
        "Unit Price",
        "Quantity",
        "Total Price",
        "Calculated Total",
        "Difference"
    ]
]
```

```python
price_error_percentage = (
    (df["Price Check"] == "Review").mean() * 100
)

price_error_percentage
```

```python
df.drop(
    columns=["Calculated Total"],
    inplace=True
)
```

```python
df[
    [
        "Age",
        "Unit Price",
        "Quantity",
        "Total Price",
        "Shipping Fee"
    ]
].describe()
```

```python
print(
    "Negative Unit Price:",
    (df["Unit Price"] < 0).sum()
)

print(
    "Invalid Quantity:",
    (df["Quantity"] <= 0).sum()
)

print(
    "Negative Total Price:",
    (df["Total Price"] < 0).sum()
)

print(
    "Negative Shipping Fee:",
    (df["Shipping Fee"] < 0).sum()
)
```

```python
print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())
```

```python
print(df["Gender"].unique())
print(df["Region"].unique())
print(df["Product Name"].unique())
print(df["Category"].unique())
print(df["Shipping Status"].unique())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df.info()
```

```python
df["Year"] = df["Order Date"].dt.year
```

```python
df["Month"] = df["Order Date"].dt.month
```

```python
df["Month Name"] = df["Order Date"].dt.month_name()
```

```python
df["Quarter"] = (
    "Q" + df["Order Date"].dt.quarter.astype(str)
)
```

```python
df["Year Month"] = (
    df["Order Date"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
bins = [0, 25, 35, 45, 55, 100]

labels = [
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56+"
]

df["Age Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)
```

```python
df["Age Group"].value_counts()
```

```python
df["Delivered Flag"] = (
    df["Shipping Status"] == "Delivered"
).astype(int)
```

```python
df["Returned Flag"] = (
    df["Shipping Status"] == "Returned"
).astype(int)
```

```python
df["In Transit Flag"] = (
    df["Shipping Status"] == "In Transit"
).astype(int)
```

```python
df[
    [
        "Shipping Status",
        "Delivered Flag",
        "Returned Flag",
        "In Transit Flag"
    ]
].head(10)
```

```python
df["Revenue"] = df["Total Price"]
```

```python
df[
    ["Unit Price", "Quantity", "Total Price", "Revenue"]
].head()
```

```python
df["Order Value Segment"] = pd.cut(
    df["Revenue"],
    bins=[0, 500, 1000, 2500, float("inf")],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
df["Order Value Segment"].value_counts()
```
